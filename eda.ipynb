```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.decomposition import PCA

# Configurar estilos de gráficos
sns.set(style='whitegrid', palette='muted', font_scale=1.2)
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 6)

```

```python
# Definir la ruta del archivo CSV
ruta_csv = os.path.join('resultados', 'matriz_3D_aplanada_dataframe.csv')

# Cargar el DataFrame
df = pd.read_csv(ruta_csv)

print(f'Dimensiones del DataFrame: {df.shape}')
df.head()
```

```python
# Información general del DataFrame
df.info()

# Estadísticas descriptivas
display(df.describe())

# Buscar valores perdidos
print('Valores perdidos por columna:')
print(df.isnull().sum())
```

```python
columnas = df.columns

# Histogramas
for col in columnas:
    plt.figure()
    sns.histplot(df[col], kde=True, bins=50, color='steelblue')
    plt.title(f'Histograma de {col}')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')
    plt.show()

# Boxplots
for col in columnas:
    plt.figure()
    sns.boxplot(x=df[col], color='lightgreen')
    plt.title(f'Boxplot de {col}')
    plt.xlabel(col)
    plt.show()
```

```python
# Matriz de correlación
corr = df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Mapa de Calor de la Matriz de Correlación')
plt.show()

# Gráfico de dispersión (pairplot)
sns.pairplot(df, diag_kind='kde', plot_kws={'alpha': 0.3})
plt.suptitle('Pairplot de las variables', y=1.02)
plt.show()
```

```python
# Número de archivos
n_archivos = 6
n_filas_archivo = int(df.shape[0] / n_archivos)

# Crear una columna 'archivo' para identificar a qué archivo pertenece cada fila
lista_archivos = []
for i in range(n_archivos):
    lista_archivos.extend([i + 1] * n_filas_archivo)

df['archivo'] = lista_archivos

print(df['archivo'].value_counts())

# Ejemplo de boxplot segmentado por archivo para la columna 'Ch1'
plt.figure(figsize=(10,6))
sns.boxplot(x='archivo', y='Ch1', data=df, palette='Set3')
plt.title('Boxplot de Ch1 segmentado por Archivo')
plt.xlabel('Archivo')
plt.ylabel('Ch1')
plt.show()

```

```python
# Seleccionar solo las columnas originales (asumiendo que la columna 'archivo' es la última)
columnas_datos = df.columns.drop('archivo')
X = df[columnas_datos].values

# Aplicar PCA
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X)

print('Varianza explicada por cada componente:', pca.explained_variance_ratio_)

# Crear un DataFrame con los componentes principales
df_pca = pd.DataFrame(data=X_pca, columns=['PC1', 'PC2'])
df_pca['archivo'] = df['archivo']

# Scatter plot de PCA
plt.figure(figsize=(10,6))
sns.scatterplot(x='PC1', y='PC2', hue='archivo', palette='Set1', data=df_pca, alpha=0.3)
plt.title('Análisis de Componentes Principales (PCA)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.legend(title='Archivo')
plt.show()
```
